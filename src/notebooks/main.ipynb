````python
import pandas as pd
from pysus import SINAN

# 1) Carrega o catálogo de arquivos disponíveis no FTP do DATASUS
sinan = SINAN().load()

# 2) Escolha o(s) ano(s). Ex: 2009..2012
years = [2009]  # troque para [2009, 2010, 2011, 2012] se quiser vários

# 3) Lista arquivos do agravo VIOL (Violência doméstica/sexual/outras)
files = sinan.get_files(dis_code="VIOL", year=years)
print("Arquivos encontrados:", [f.name for f in files])
````

````python
# 1) Carrega o catálogo de arquivos disponíveis no FTP do DATASUS
sinan = SINAN().load()

# 2) Escolha o(s) ano(s). Ex: 2009..2012
years = [2009]  # troque para [2009, 2010, 2011, 2012] se quiser vários

# 3) Lista arquivos do agravo VIOL (Violência doméstica/sexual/outras)
files = sinan.get_files(dis_code="VIOL", year=years)
print("Arquivos encontrados:", [f.name for f in files])

# 4) Faz download (gera .parquet local) e lê no pandas
dfs = []
for f in files:
    parquet_path = f.download(local_dir="/content/sinan_viol_parquets")
    df = parquet_path.to_dataframe()
    df["__ARQUIVO"] = f.name  # só pra rastrear origem
    dfs.append(df)

df_all = pd.concat(dfs, ignore_index=True)
df_all.head()

# 5) Salva em Parquet (rápido) e/ou CSV (maior)
df_all.to_parquet("/content/sinan_viol.parquet", index=False)
df_all.to_csv("/content/sinan_viol.csv", index=False)

print("OK! Arquivos gerados em /content:")
print("- /content/sinan_viol.parquet")
print("- /content/sinan_viol.csv")
````
