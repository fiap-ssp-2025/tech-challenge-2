## 1. OBTER DATASETS E CONCATENAÇÃO

Objetivo do código abaixo é obter os dados do SUS mencionados acima. Estes dados vêm originalmente no format `DBC` então vamos criar uma pequena pipeline para convertê-los em parquet. 

Parquet é um formato colunas, similar a de um banco de dados. Se estiver usando o VS Code, sugiro que baixe o `Parquet Visualizer` para poder visualizar o conteúdo dos arquivos. 

Do arquivo DBC, vários arquivos parquet são gerados então faremos também a concatenização dos parquets e conversão para DataFrame do Pandas para poder manipular o dataset.


````python
## Base path (raiz do projeto)
from pathlib import Path

# Assumindo que o Notebook/Jupyter está sendo executado em /src/notebooks
BASE_PATH = Path.cwd().resolve().parent.parent
print(f"📁 Base path: {BASE_PATH.absolute()}")
````

````python
import pandas as pd
from pysus import SINAN


sinan = SINAN().load()
years = [2024]  

# 3) Lista arquivos do agravo VIOL (Violência doméstica/sexual/outras)
dbcFiles = sinan.get_files(dis_code="VIOL", year=years)

print(f"Arquivos encontrados: {dbcFiles}")


dfs = []
for f in dbcFiles:
    parquet_path = f.download(local_dir=f"{BASE_PATH}/data/raw")
    df = parquet_path.to_dataframe()
    df["__ARQUIVO"] = f.name  # só pra rastrear origem
    dfs.append(df)

df_all = pd.concat(dfs, ignore_index=True)
df_all.head()
df_all.columns
df_all.dtypes
df_all.shape
df_all.info()
df_all.describe()
df_all.isnull().sum()
df_all.isnull().sum() / len(df_all)
````
