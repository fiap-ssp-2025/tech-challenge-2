Objetivo do código abaixo é obter os dados do SUS mencionados acima. Estes dados vêm originalmente no format `DBC` então vamos criar uma pequena pipeline para convertê-los em parquet.

````python
## Função para encontrar a raiz do projeto

import pandas as pd
from pysus import SINAN
from pathlib import Path

# Detecta a raiz do projeto procurando por README.md ou .git
def find_project_root():
    """Sobe diretórios até encontrar a raiz do projeto."""
    current = Path.cwd()
    
    # Procura subindo até 5 níveis
    for _ in range(5):
        # Verifica se é a raiz do projeto (tem README.md ou .git)
        if (current / "README.md").exists() or (current / ".git").exists():
            return current
        # Se estiver em src/notebooks, sobe 2 níveis
        if current.name == "notebooks" and (current.parent.parent / "README.md").exists():
            return current.parent.parent
        current = current.parent
    
    # Fallback: retorna 2 níveis acima do diretório atual
    return Path.cwd().parent.parent

BASE_PATH = find_project_root()
print(f"📁 Base path: {BASE_PATH.absolute()}")
````

````python
import pandas as pd
from pysus import SINAN
from dbfread import DBF

# 1) Carrega o catálogo de arquivos disponíveis no FTP do DATASUS
sinan = SINAN().load()

# 2) Escolha o(s) ano(s). Ex: 2009..2012
years = [2024]  # troque para [2009, 2010, 2011, 2012] se quiser vários

# 3) Lista arquivos do agravo VIOL (Violência doméstica/sexual/outras)
dbcFiles = sinan.get_files(dis_code="VIOL", year=years)
# Printa os arquivos encontrados com a extensão do arquivo
print(f"Arquivos encontrados: {dbcFiles}")

# 2) Usa o método download() que já converte DBC -> Parquet
# 4) Faz download (gera .parquet local) e lê no pandas
dfs = []
for f in dbcFiles:
    parquet_path = f.download(local_dir=f"{BASE_PATH}/data/raw")
    df = parquet_path.to_dataframe()
    df["__ARQUIVO"] = f.name  # só pra rastrear origem
    dfs.append(df)

df_all = pd.concat(dfs, ignore_index=True)
df_all.head()
df_all.columns
df_all.dtypes
df_all.shape
df_all.info()
df_all.describe()
df_all.isnull().sum()
df_all.isnull().sum() / len(df_all)
````
